## Q6. Employee Dataset

```
employees = {
    "Employee_ID":[101,102,103,104,105],
    "Name":["Alice","Bob","Charlie","Diana","Edward"],
    "Department":["HR","IT","IT","Marketing","Sales"],
    "Age":[29,34,41,28,38],
    "Salary":[50000,70000,65000,55000,60000],
    "Years_of_Experience":[4,8,10,3,12],
    "Joining_Date":["2020-03-15","2017-07-19","2013-06-01",
                    "2021-02-10","2010-11-25"],
    "Gender":["Female","Male","Male","Female","Male"],
    "Bonus":[5000,7000,6000,4500,5000],
    "Rating":[4.5,4.0,3.8,4.7,3.5]
}

emp = pd.DataFrame(employees)
emp
```

```
# a) Shape
emp.shape
```

```
# b) Summary including datatype and non-null count
emp.info()
```

```
# c) Descriptive statistics
emp.describe()
```

```
# d) First 5 rows and last 3 rows

print(emp.head())
print(emp.tail(3))
```

```
# e) Statistics

print("Average Salary:", emp["Salary"].mean())
print("Total Bonus:", emp["Bonus"].sum())
print("Youngest Employee Age:", emp["Age"].min())
print("Highest Performance Rating:", emp["Rating"].max())
```

```
# f) Sort by Salary descending

emp.sort_values(by="Salary", ascending=False)
```

```
# g) Add Performance category

def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp["Performance"] = emp["Rating"].apply(performance)
emp
```

```
# h) Identify missing values

emp.isnull().sum()
```

```
# i) Rename Employee_ID to ID

emp.rename(columns={"Employee_ID":"ID"}, inplace=True)
emp
```

```
# j) Employees with more than 5 years experience

print(emp[emp["Years_of_Experience"] > 5])

# Employees belonging to IT department
print(emp[emp["Department"] == "IT"])
```

```
# k) Add Tax column (10% deduction)

emp["Tax"] = emp["Salary"] * 0.10
emp
```

```
# l) Save modified DataFrame

emp.to_csv("modified_employees.csv", index=False)

print("File saved successfully")
```
